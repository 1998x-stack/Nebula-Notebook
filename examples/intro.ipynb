## Generate a sine wave
This plot is rendered in the Notebook using Jupyter’s `image/png` output.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 4*np.pi, 300)
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, np.sin(x), label="sin(x)")
ax.plot(x, np.cos(x), label="cos(x)", alpha=0.75)
ax.legend()
ax.set_title("Nebula Notebook — interactive output")
plt.show()
```
